```python
theta1, theta2, theta3, theta4 = symbols('theta_1, theta_2, theta_3, theta_4')
l1, l2, l3, l4 = symbols('l_1, l_2, l_3, l_4')

N = ReferenceFrame('N')
A = N.orientnew('N', 'Axis', (N.z, theta1))
B = A.orientnew('N', 'Axis', (N.z, theta2))
C = B.orientnew('N', 'Axis', (N.z, theta3))
D = N.orientnew('N', 'Axis', (N.z, theta4))

pos = l2*B.x + l3*C.x + l4*D.x - l1*A.x
v_pos = [pos.dot(N.x), pos.dot(N.y)]

st_max = {l1: 0.195, theta2: np.deg2rad(145), theta4: theta3+np.pi}
st_min = {l1: 0.195, theta2: np.deg2rad(35), theta4: theta3}

eq_list = [eq.subs(st_max) for eq in v_pos]
eq_list = eq_list + [eq.subs(st_min) for eq in v_pos]

eq_list
```

```python
eq_fun = lambdify([l2, l3, l4, theta1, theta3], eq_list)
sln = least_squares(lambda x: eq_fun(*x), [0.1, 0.1, 0.1, np.deg2rad(45), np.deg2rad(45)])
sln.x
```
